```python
import pandas as pd
import numpy as np

# Load the original UCI Landsat training and test datasets
train_data = pd.read_csv("sat.trn", sep=r"\s+", header=None)
test_data = pd.read_csv("sat.tst", sep=r"\s+", header=None)

print("Training data shape:", train_data.shape)
print("Test data shape:", test_data.shape)

train_data.head()
```

```python
# Assign meaningful column names
feature_names = [f"Feature_{i}" for i in range(1, 37)]
column_names = feature_names + ["Class"]

train_data.columns = column_names
test_data.columns = column_names

print("Total training instances:", len(train_data))
print("Total test instances:", len(test_data))
print("Number of input features:", len(feature_names))

print("\nClass labels in training data:")
print(sorted(train_data["Class"].unique()))

print("\nMissing values in training data:", train_data.isnull().sum().sum())
print("Missing values in test data:", test_data.isnull().sum().sum())

train_data.head()
```

```python
# Check class distribution in the training dataset
class_distribution = train_data["Class"].value_counts().sort_index()

print("Training Class Distribution:")
print(class_distribution)

print("\nTraining Class Distribution (%):")
print((class_distribution / len(train_data) * 100).round(2))
```

```python
# Separate input features and target class

X_train = train_data.drop("Class", axis=1)
y_train = train_data["Class"]

X_test = test_data.drop("Class", axis=1)
y_test = test_data["Class"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize the features for scale-sensitive models such as
# Logistic Regression, KNN, and Support Vector Machine.
scaler = StandardScaler()

# Fit only on the training data to avoid data leakage.
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same fitted scaler to the test data.
X_test_scaled = scaler.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)

print("\nMean of scaled training features:", round(X_train_scaled.mean(), 4))
print("Standard deviation:", round(X_train_scaled.std(), 4))
```

```python
from sklearn.linear_model import LogisticRegression

# Model 1: Logistic Regression
# Train the classifier using the standardized training features.
logistic_model = LogisticRegression(max_iter=2000, random_state=42)

logistic_model.fit(X_train_scaled, y_train)

# Generate class predictions and class probabilities on the test dataset.
logistic_predictions = logistic_model.predict(X_test_scaled)
logistic_probabilities = logistic_model.predict_proba(X_test_scaled)

print("Logistic Regression training completed successfully.")
print("Number of test predictions:", len(logistic_predictions))
print("Classes learned by model:", logistic_model.classes_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Calculate the six evaluation metrics required in the assignment
logistic_accuracy = accuracy_score(y_test, logistic_predictions)

logistic_auc = roc_auc_score(
    y_test,
    logistic_probabilities,
    multi_class="ovr",
    average="weighted",
    labels=logistic_model.classes_
)

logistic_precision = precision_score(
    y_test, logistic_predictions, average="weighted"
)

logistic_recall = recall_score(
    y_test, logistic_predictions, average="weighted"
)

logistic_f1 = f1_score(
    y_test, logistic_predictions, average="weighted"
)

logistic_mcc = matthews_corrcoef(
    y_test, logistic_predictions
)

print("Logistic Regression Evaluation Metrics")
print("--------------------------------------")
print("Accuracy :", round(logistic_accuracy, 4))
print("AUC      :", round(logistic_auc, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("F1 Score :", round(logistic_f1, 4))
print("MCC Score:", round(logistic_mcc, 4))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Model 2: Decision Tree Classifier
# Decision Trees do not require standardized features,
# so the original training features are used.

decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
decision_tree_model.fit(X_train, y_train)

# Generate predictions and class probabilities on the test dataset
decision_tree_predictions = decision_tree_model.predict(X_test)
decision_tree_probabilities = decision_tree_model.predict_proba(X_test)

print("Decision Tree training completed successfully.")
print("Number of test predictions:", len(decision_tree_predictions))
print("Classes learned by model:", decision_tree_model.classes_)
```

```python
# Calculate the six required evaluation metrics for Decision Tree

decision_tree_accuracy = accuracy_score(
    y_test, decision_tree_predictions
)

decision_tree_auc = roc_auc_score(
    y_test,
    decision_tree_probabilities,
    multi_class="ovr",
    average="weighted",
    labels=decision_tree_model.classes_
)

decision_tree_precision = precision_score(
    y_test, decision_tree_predictions, average="weighted"
)

decision_tree_recall = recall_score(
    y_test, decision_tree_predictions, average="weighted"
)

decision_tree_f1 = f1_score(
    y_test, decision_tree_predictions, average="weighted"
)

decision_tree_mcc = matthews_corrcoef(
    y_test, decision_tree_predictions
)

print("Decision Tree Evaluation Metrics")
print("--------------------------------")
print("Accuracy :", round(decision_tree_accuracy, 4))
print("AUC      :", round(decision_tree_auc, 4))
print("Precision:", round(decision_tree_precision, 4))
print("Recall   :", round(decision_tree_recall, 4))
print("F1 Score :", round(decision_tree_f1, 4))
print("MCC Score:", round(decision_tree_mcc, 4))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Model 3: K-Nearest Neighbor (KNN) Classifier
# KNN is distance-based, therefore the standardized features are used.

knn_model = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_model.fit(X_train_scaled, y_train)

# Generate predictions and class probabilities on the test dataset
knn_predictions = knn_model.predict(X_test_scaled)
knn_probabilities = knn_model.predict_proba(X_test_scaled)

print("KNN training completed successfully.")
print("Number of test predictions:", len(knn_predictions))
print("Classes learned by model:", knn_model.classes_)
```

```python
# Calculate the six required evaluation metrics for KNN

knn_accuracy = accuracy_score(
    y_test, knn_predictions
)

knn_auc = roc_auc_score(
    y_test,
    knn_probabilities,
    multi_class="ovr",
    average="weighted",
    labels=knn_model.classes_
)

knn_precision = precision_score(
    y_test, knn_predictions, average="weighted"
)

knn_recall = recall_score(
    y_test, knn_predictions, average="weighted"
)

knn_f1 = f1_score(
    y_test, knn_predictions, average="weighted"
)

knn_mcc = matthews_corrcoef(
    y_test, knn_predictions
)

print("KNN Evaluation Metrics")
print("----------------------")
print("Accuracy :", round(knn_accuracy, 4))
print("AUC      :", round(knn_auc, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("F1 Score :", round(knn_f1, 4))
print("MCC Score:", round(knn_mcc, 4))
```

```python
from sklearn.naive_bayes import GaussianNB

# Model 4: Gaussian Naive Bayes Classifier
# Gaussian Naive Bayes is suitable for numerical input features.

naive_bayes_model = GaussianNB()

# Train the model using the original training features
naive_bayes_model.fit(X_train, y_train)

# Generate predictions and class probabilities on the test dataset
naive_bayes_predictions = naive_bayes_model.predict(X_test)
naive_bayes_probabilities = naive_bayes_model.predict_proba(X_test)

print("Naive Bayes training completed successfully.")
print("Number of test predictions:", len(naive_bayes_predictions))
print("Classes learned by model:", naive_bayes_model.classes_)
```

```python
# Calculate the six required evaluation metrics for Gaussian Naive Bayes

naive_bayes_accuracy = accuracy_score(
    y_test, naive_bayes_predictions
)

naive_bayes_auc = roc_auc_score(
    y_test,
    naive_bayes_probabilities,
    multi_class="ovr",
    average="weighted",
    labels=naive_bayes_model.classes_
)

naive_bayes_precision = precision_score(
    y_test, naive_bayes_predictions, average="weighted"
)

naive_bayes_recall = recall_score(
    y_test, naive_bayes_predictions, average="weighted"
)

naive_bayes_f1 = f1_score(
    y_test, naive_bayes_predictions, average="weighted"
)

naive_bayes_mcc = matthews_corrcoef(
    y_test, naive_bayes_predictions
)

print("Naive Bayes Evaluation Metrics")
print("------------------------------")
print("Accuracy :", round(naive_bayes_accuracy, 4))
print("AUC      :", round(naive_bayes_auc, 4))
print("Precision:", round(naive_bayes_precision, 4))
print("Recall   :", round(naive_bayes_recall, 4))
print("F1 Score :", round(naive_bayes_f1, 4))
print("MCC Score:", round(naive_bayes_mcc, 4))
```

```python
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV

# Model 5: Support Vector Machine (SVM)
# SVM is scale-sensitive, so standardized features are used.
# Probability calibration is applied to obtain probabilities required for AUC.

base_svm = SVC(
    kernel="rbf"
)

svm_model = CalibratedClassifierCV(
    base_svm,
    method="sigmoid",
    cv=5,
    ensemble=False
)

# Train the calibrated SVM model
svm_model.fit(X_train_scaled, y_train)

# Generate predictions and calibrated class probabilities
svm_predictions = svm_model.predict(X_test_scaled)
svm_probabilities = svm_model.predict_proba(X_test_scaled)

print("SVM training completed successfully.")
print("Number of test predictions:", len(svm_predictions))
print("Classes learned by model:", svm_model.classes_)
```

```python
# Calculate the six required evaluation metrics for SVM

svm_accuracy = accuracy_score(
    y_test, svm_predictions
)

svm_auc = roc_auc_score(
    y_test,
    svm_probabilities,
    multi_class="ovr",
    average="weighted",
    labels=svm_model.classes_
)

svm_precision = precision_score(
    y_test, svm_predictions, average="weighted"
)

svm_recall = recall_score(
    y_test, svm_predictions, average="weighted"
)

svm_f1 = f1_score(
    y_test, svm_predictions, average="weighted"
)

svm_mcc = matthews_corrcoef(
    y_test, svm_predictions
)

print("Support Vector Machine Evaluation Metrics")
print("-----------------------------------------")
print("Accuracy :", round(svm_accuracy, 4))
print("AUC      :", round(svm_auc, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("MCC Score:", round(svm_mcc, 4))
```

```python
# Create the model comparison table required for the assignment

results = pd.DataFrame({
    "ML Model Name": [
        "Logistic Regression",
        "Decision Tree",
        "KNN",
        "Naive Bayes",
        "Support Vector Machine"
    ],
    "Accuracy": [
        logistic_accuracy,
        decision_tree_accuracy,
        knn_accuracy,
        naive_bayes_accuracy,
        svm_accuracy
    ],
    "AUC": [
        logistic_auc,
        decision_tree_auc,
        knn_auc,
        naive_bayes_auc,
        svm_auc
    ],
    "Precision": [
        logistic_precision,
        decision_tree_precision,
        knn_precision,
        naive_bayes_precision,
        svm_precision
    ],
    "Recall": [
        logistic_recall,
        decision_tree_recall,
        knn_recall,
        naive_bayes_recall,
        svm_recall
    ],
    "F1": [
        logistic_f1,
        decision_tree_f1,
        knn_f1,
        naive_bayes_f1,
        svm_f1
    ],
    "MCC": [
        logistic_mcc,
        decision_tree_mcc,
        knn_mcc,
        naive_bayes_mcc,
        svm_mcc
    ]
})

results = results.round(4)

print("Comparison of Classification Models")
display(results)
```
